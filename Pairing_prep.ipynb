```python
#Create Dataframe For test RTI Data
new_df=df[['time','gate']]
new=pd.concat([new_df,new_df],ignore_index=True)
new[2] = np.where(new.index >= 14, 8, 9)
new[3] = 10
new
```

```python
def adjusted_norm(dataframe):#Should workshop the name
    
    threshold = 7 # The Largest norm which will still be accepted as a correct value
    col = 2
    norm_list=['time','ADSB_plane']

    while col <= dataframe.shape[1]-1:
        norm=np.linalg.norm(dataframe['gate']-dataframe[col])
        norm_list.append(norm)
        col +=1
    matched_norm=min(norm_list[2:])

    if matched_norm > threshold:
        return "None",norm_list
    
    matched_col=norm_list.index(matched_norm)
    return matched_col, norm_list
```
