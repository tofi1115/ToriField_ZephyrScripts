```python
from opensky_api import OpenSkyApi
api = OpenSkyApi("Tofi1115","becsuv-8cohTa-meqsaw")
```

```python
# Latitude and longitude of transmiter and reciever
boulder_lat=40.0190
boulder_lon=-105.2747

parker_lat=39.5186
parker_lon=-104.7614
```

```python
plane_time=1754968222 #1754631492 #1754631462 #Last time of plane above colorado area
min_time=plane_time-245 #Set min time to 30s before 

Plane_states=[]

import time as tm
```

```python
time=min_time
while time<=plane_time:
    states = api.get_states(time_secs=time,bbox=(parker_lat, boulder_lat, boulder_lon, parker_lon))
    print(states)
    if states!=None:
        for s in states.states:
                vector=[s.callsign, s.latitude,s.longitude,s.category,s.on_ground,s.time_position,s.baro_altitude,time]
                Plane_states.append(vector)
                print(vector)    
    time+=10
    tm.sleep(10)
```

```python
max_time=240
t=0
while t<=max_time:
    states = api.get_states(time_secs=0,bbox=(parker_lat, boulder_lat, boulder_lon, parker_lon))
    print(states)
    if states!=None:
        for s in states.states:
                vector=[s.callsign, s.latitude,s.longitude,s.category,s.on_ground,s.time_position,s.baro_altitude,t]
                Plane_states.append(vector)
                #print(vector)    
    print(t)
    t+=5
    tm.sleep(5)
```
